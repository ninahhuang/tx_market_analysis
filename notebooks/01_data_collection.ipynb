## 1. Load Zillow Home Value Index Data

Zillow Home Value Index (ZHVI) represents the typical home value within a geographic area. The raw city-level dataset contains one row per geographic region and separate columns for each monthly observation.

The raw dataset is preserved unchanged in `data/raw/`.

```python
zhvi_raw = pd.read_csv(ZILLOW_FILE)

print("Dataset shape:", zhvi_raw.shape)

zhvi_raw.head()
```

```python
# inspect columns

print("Number of columns:", len(zhvi_raw.columns))

zhvi_raw.columns.tolist()[:30]
```

```python
zhvi_raw.info()
```

## 2. Define Target Markets

The initial analysis includes eight North Dallas-area cities representing a mix of emerging, rapidly growing, established-growth, and mature suburban housing markets.

```python
# target market definitions

TARGET_CITIES = [
    "Anna",
    "Celina",
    "Frisco",
    "McKinney",
    "Melissa",
    "Plano",
    "Princeton",
    "Prosper",
]

TARGET_STATE = "TX"
START_DATE = "2018-01-01"

print("Number of target cities:", len(TARGET_CITIES))
print(TARGET_CITIES)
```

```python
# identify target city records

target_check = zhvi_raw[
    zhvi_raw["RegionName"].isin(TARGET_CITIES)
].copy()

metadata_cols = [
    col for col in [
        "RegionID",
        "RegionName",
        "RegionType",
        "StateName",
        "State",
        "Metro",
        "CountyName"
    ]
    if col in target_check.columns
]

target_check[metadata_cols].sort_values("RegionName")
```

```python
[col for col in zhvi_raw.columns if "state" in col.lower()]
```

```python
for col in zhvi_raw.columns:
    if "state" in col.lower():
        print(f"\n{col}:")
        print(zhvi_raw[col].dropna().unique()[:20])
```

```python
# filter target TX cities

zhvi_target = zhvi_raw[
    (zhvi_raw["RegionName"].isin(TARGET_CITIES))
    & (zhvi_raw["StateName"] == "TX")
].copy()

print("Rows found:", len(zhvi_target))

zhvi_target[metadata_cols].sort_values("RegionName")
```

```python
# validate city coverage

cities_found = sorted(zhvi_target["RegionName"].unique())
cities_missing = sorted(set(TARGET_CITIES) - set(cities_found))

print("Cities found:")
for city in cities_found:
    print("-", city)

print("\nCities missing:")
if cities_missing:
    for city in cities_missing:
        print("-", city)
else:
    print("None")
```

```python
# identify monthly date columns

date_cols = []

for col in zhvi_target.columns:
    try:
        pd.to_datetime(col, format="%Y-%m-%d")
        date_cols.append(col)
    except (ValueError, TypeError):
        pass

print("Number of monthly date columns:", len(date_cols))
print("First five date columns:", date_cols[:5])
print("Last five date columns:", date_cols[-5:])
```

## 3. Reshape Zillow Data

The original Zillow dataset stores individual months as separate columns. For time-series analysis, the data are converted from wide format to long format, where each row represents one city-month observation.

```python
# reshape wide data to long format

zhvi_long = zhvi_target.melt(
    id_vars=["RegionName"],
    value_vars=date_cols,
    var_name="Date",
    value_name="ZHVI"
)

zhvi_long = zhvi_long.rename(
    columns={"RegionName": "City"}
)

zhvi_long.head(10)
```

```python
# convert date column

zhvi_long["Date"] = pd.to_datetime(zhvi_long["Date"])

print(zhvi_long.dtypes)
```

```python
# filter study period to 2018-2026

zhvi_long = zhvi_long[
    zhvi_long["Date"] >= pd.Timestamp(START_DATE)
].copy()

zhvi_long = zhvi_long.sort_values(
    ["City", "Date"]
).reset_index(drop=True)

print("Dataset shape after date filter:", zhvi_long.shape)

print(
    "Study period:",
    zhvi_long["Date"].min().date(),
    "to",
    zhvi_long["Date"].max().date()
)

zhvi_long.head()
```

```python
# create time variables

zhvi_long["Year"] = zhvi_long["Date"].dt.year
zhvi_long["Month"] = zhvi_long["Date"].dt.month
zhvi_long["YearMonth"] = zhvi_long["Date"].dt.to_period("M").astype(str)

zhvi_long.head()
```

```python
# duplicate check

duplicate_count = zhvi_long.duplicated(
    subset=["City", "Date"]
).sum()

print("Duplicate city-month rows:", duplicate_count)
```

```python
# missing value check

missing_by_city = (
    zhvi_long
    .groupby("City")["ZHVI"]
    .agg(
        Observations="size",
        Missing=lambda x: x.isna().sum(),
        Available=lambda x: x.notna().sum()
    )
)

missing_by_city
```

```python
# data coverage by city

coverage = (
    zhvi_long
    .dropna(subset=["ZHVI"])
    .groupby("City")
    .agg(
        First_Date=("Date", "min"),
        Last_Date=("Date", "max"),
        Observations=("ZHVI", "count")
    )
    .reset_index()
)

coverage
```

```python
# sanity check; summary statistics
zhvi_long.groupby("City")["ZHVI"].agg(
    ["count", "min", "median", "max"]
).round(0)
```

```python
# latest available ZHVI by city

latest_zhvi = (
    zhvi_long
    .dropna(subset=["ZHVI"])
    .sort_values("Date")
    .groupby("City", as_index=False)
    .tail(1)
    [["City", "Date", "ZHVI"]]
    .sort_values("ZHVI", ascending=False)
)
```

```python
# display results
latest_zhvi_display = latest_zhvi.copy()

latest_zhvi_display["ZHVI"] = latest_zhvi_display["ZHVI"].map(
    lambda x: f"${x:,.0f}"
)

latest_zhvi_display
```

```python
# first available observation from 2018

starting_zhvi = (
    zhvi_long
    .dropna(subset=["ZHVI"])
    .sort_values("Date")
    .groupby("City", as_index=False)
    .first()
    [["City", "Date", "ZHVI"]]
)

starting_zhvi
```

## 4. Export Processed Zillow Dataset

The validated Zillow dataset is exported in long format for use in subsequent exploratory analysis and modeling notebooks.

Each row represents one city-month observation.

```python
# export cleaned dataset

OUTPUT_FILE = PROCESSED_DATA_DIR / "tx_zhvi_monthly.csv"

zhvi_long.to_csv(
    OUTPUT_FILE,
    index=False
)

print("Saved:", OUTPUT_FILE)
print("Rows exported:", len(zhvi_long))
```

```python
# verify

zhvi_check = pd.read_csv(
    OUTPUT_FILE,
    parse_dates=["Date"]
)

print("Reloaded shape:", zhvi_check.shape)
print("Cities:", sorted(zhvi_check["City"].unique()))

zhvi_check.head()
```
